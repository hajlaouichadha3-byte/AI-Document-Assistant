```python

def repondre_question(question, texte):
    question = question.lower()
    texte = texte.lower()

    if "python" in question:
        if "python" in texte:
            return "Python est un langage populaire pour la Data Science."

    if "machine learning" in question:
        if "machine learning" in texte:
            return "Le Machine Learning permet aux ordinateurs d'apprendre à partir des données."

    if "data science" in question:
        if "data science" in texte:
            return "La Data Science utilise les données pour extraire des informations utiles."

    return "Je ne trouve pas la réponse dans le document."

question = "Qu'est-ce que Python ?"

print("Question :", question)
print("Réponse :", repondre_question(question, texte))
```

```python

def rechercher_dans_document(question, texte):
    phrases = texte.split(".")
    mots_question = set(question.lower().split())

    meilleur_score = 0
    meilleure_phrase = ""

    for phrase in phrases:
        mots_phrase = set(phrase.lower().split())
        score = len(mots_question & mots_phrase)

        if score > meilleur_score:
            meilleur_score = score
            meilleure_phrase = phrase.strip()

    if meilleure_phrase:
        return meilleure_phrase

    return "Aucune information pertinente trouvée."

question = input("Posez votre question : ")
print("\nRéponse :", rechercher_dans_document(question, texte))
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def rechercher_intelligemment(question, texte):
    phrases = [
        phrase.strip()
        for phrase in texte.split(".")
        if phrase.strip()
    ]

    if not phrases:
        return "Le document ne contient pas de texte exploitable."

    vectorizer = TfidfVectorizer()
    matrice = vectorizer.fit_transform(phrases + [question])

    similarites = cosine_similarity(
        matrice[-1],
        matrice[:-1]
    )[0]

    meilleur_index = similarites.argmax()

    if similarites[meilleur_index] == 0:
        return "Je n'ai pas trouvé d'information pertinente."

    return phrases[meilleur_index]


question = input("Posez votre question : ")
print("\nRéponse :", rechercher_intelligemment(question, texte))
```

```python

questions = [
    "What is Data Science?",
    "Which language is popular for Data Science?",
    "How do computers learn from data?"
]

for question in questions:
    reponse = rechercher_intelligemment(question, texte)

    print("Question :", question)
    print("Réponse :", reponse)
    print("-" * 50)
```

```python

def rechercher_avec_score(question, texte):
    phrases = [
        phrase.strip()
        for phrase in texte.split(".")
        if phrase.strip()
    ]

    if not phrases:
        return None, 0.0

    vectorizer = TfidfVectorizer()
    matrice = vectorizer.fit_transform(phrases + [question])

    scores = cosine_similarity(
        matrice[-1],
        matrice[:-1]
    )[0]

    meilleur_index = scores.argmax()
    meilleure_phrase = phrases[meilleur_index]
    meilleur_score = float(scores[meilleur_index])

    if meilleur_score == 0:
        return "Aucune information pertinente trouvée.", 0.0

    return meilleure_phrase, meilleur_score


question = "Python is a popular language for Data Science"

reponse, score = rechercher_avec_score(question, texte)

print("\nRéponse :", reponse)
print(f"Score de similarité : {score:.2f}")
```
